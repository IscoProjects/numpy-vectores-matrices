## 1.1 Anatomía del `ndarray`: Rank, Axis, Shape, Size, Index y Value

A diferencia de una lista de Python (que almacena punteros dispersos a objetos heterogéneos en memoria), un **`ndarray`** (*N-dimensional array*) es un bloque continuo de memoria donde todos los elementos comparten exactamente el mismo tipo de dato (`dtype`).

### Conceptos Estructurales Clave
* **Rank (`ndim`):** Número total de dimensiones o ejes del arreglo. Un vector tiene `ndim = 1`; una matriz tiene `ndim = 2`; un cubo de datos tiene `ndim = 3`.
* **Shape (`shape`):** Tupla de enteros que indica cuántos elementos existen a lo largo de cada dimensión. Para una matriz de 3 filas y 4 columnas, `shape` es `(3, 4)`.
* **Axis (`axis`):** Dirección a lo largo de la cual se ejecutan las operaciones:
  * `axis=0`: Eje vertical (recorre las **filas** de arriba hacia abajo).
  * `axis=1`: Eje horizontal (recorre las **columnas** de izquierda a derecha).
* **Size (`size`):** Cantidad total de elementos almacenados (equivale al producto de los valores de `shape`).
* **Index y Value:** El *index* es la coordenada posicional (siempre inicia en `0`), y el *value* es el dato escalar almacenado en esa celda.

### Sintaxis Mínima

````python
# 1. Vector 1D: Registro de temperaturas diarias durante 10 días (°C)
temperaturas = np.array([22.5, 23.1, 21.8, 24.0, 25.2, 24.8, 23.9, 22.0, 21.5, 23.4])

# 2. Matriz 2D (3x4): Calificaciones de 3 estudiantes en 4 evaluaciones
calificaciones = np.array([
    [8.5, 9.0, 7.8, 9.2],
    [6.5, 7.0, 8.1, 7.5],
    [9.5, 9.8, 9.2, 10.0]
])

# 3. Arreglo 3D (2x3x4): Calificaciones de 2 paralelos (bloques), 3 estudiantes, 4 evaluaciones
campus_notas = np.array([
    [[8.5, 9.0, 7.8, 9.2], [6.5, 7.0, 8.1, 7.5], [9.5, 9.8, 9.2, 10.0]],
    [[7.0, 7.5, 8.0, 8.2], [8.8, 8.5, 9.0, 8.9], [6.0, 6.8, 7.2, 7.0]]
])

def inspeccionar(nombre, arr):
    print(f"--- {nombre} ---")
    print(arr)
    print(f"Rank (ndim) : {arr.ndim}D")
    print(f"Shape       : {arr.shape}")
    print(f"Size        : {arr.size} elementos")
    print(f"Tipo (dtype): {arr.dtype}\n")

inspeccionar("Vector Temperaturas", temperaturas)
inspeccionar("Matriz Calificaciones", calificaciones)
inspeccionar("Tensor 3D Campus", campus_notas)

# Acceso básico a Index y Value
print(f"Index [0] en vector -> Value: {temperaturas[0]}")
print(f"Index [2, 3] en matriz (Fila 2, Columna 3) -> Value: {calificaciones[2, 3]}")
````

:::{note} Reglas y Restricciones del `ndarray`
1. **Homogeneidad estricta:** Todos los elementos deben tener el mismo `dtype`. Si mezclas enteros y flotantes, NumPy promueve todo el arreglo al tipo más general (`float64`).
2. **Rectangularidad obligatoria:** Todas las filas de una matriz 2D deben tener exactamente la misma longitud. No existen matrices "dentadas" (*ragged arrays*) de forma nativa.
:::

:::{warning} Trampa Común: Confundir `shape` de un Vector 1D con una Matriz Fila.
Un vector de 10 elementos tiene `shape = (10,)` (una tupla de un solo elemento), **no** `(1, 10)` ni `(10, 1)`. Mientras que una matriz fila de $1 \times 10$ tiene `ndim = 2`, un vector 1D tiene `ndim = 1`.
:::

## 1.2 Creación de Vectores y Matrices: Secuencias e Inicialización

En ciencia de datos e ingeniería, rara vez escribimos matrices grandes a mano. Lo habitual es reservar bloques de memoria pre-inicializados para llenarlos posteriormente mediante algoritmos o simulaciones.

### Funciones Principales
* **`np.array(lista)`:** Convierte listas o tuplas anidadas de Python en un `ndarray`.
* **`np.zeros(shape)`:** Crea un arreglo lleno de ceros (`0.0` por defecto). Ideal para acumuladores o máscaras.
* **`np.ones(shape)`:** Crea un arreglo lleno de unos (`1.0` por defecto).
* **`np.full(shape, fill_value)`:** Crea un arreglo rellenado con un valor escalar específico definido por el usuario.
* **`np.empty(shape)`:** Reserva el espacio en memoria **sin inicializar los valores** (contiene "basura" residual de la RAM). Es ligeramente más rápido que `zeros` cuando sabes que sobrescribirás el 100% de las celdas inmediatamente.
* **`np.eye(N)`:** Crea una matriz identidad cuadrada de $N \times N$ con `1` en la diagonal principal y `0` en el resto.

### Sintaxis Mínima

````python
# Matriz de ceros de 3 filas x 5 columnas
ceros = np.zeros((3, 5))

# Matriz de unos de 4x4 con tipo entero explícito
unos_enteros = np.ones((4, 4), dtype=np.int64)

# Matriz de 3x4 inicializada con un valor constante
matriz_constante = np.full((3, 4), fill_value=20.5)

# Matriz vacía de 3x3
memoria_sin_limpiar = np.empty((3, 3))

# Matriz identidad de 4x4
identidad = np.eye(4)

print("Matriz ceros (3x5):\n", ceros)
print("\nMatriz unos (4x4):\n", unos_enteros)
print("\nMatriz full (3x4):\n", matriz_constante)
print("\nMatriz empty (3x3 - valores arbitrarios de RAM):\n", memoria_sin_limpiar)
print("\nMatriz identidad (4x4):\n", identidad)
````

:::{warning} Trampa Común: Olvidar los paréntesis internos en `shape`
Un error de sintaxis muy frecuente es escribir `np.zeros(3, 4)` en lugar de `np.zeros((3, 4))`.
El primer argumento de `zeros`, `ones`, `full` y `empty` debe ser **una tupla** `(filas, columnas)`. Si pasas dos números sueltos, NumPy interpretará el segundo número (`4`) como el `dtype` y lanzará un `TypeError`.
:::

:::{tip} Regla de Seguridad con `np.empty()`
Nunca asumas que `np.empty()` contiene ceros. Si vas a realizar sumas acumulativas (`matriz += valor`), utiliza siempre `np.zeros()`. Usa `np.empty()` únicamente cuando vayas a asignar valores directos (`matriz[i, j] = valor`) a todas las posiciones.
:::

## 1.3 Creación mediante Rangos Numéricos: `np.arange` vs. `np.linspace`

Para generar ejes de tiempo, dominios matemáticos o cuadrículas de evaluación sin escribir listas manualmente, NumPy ofrece dos generadores complementarios:

| Función | Parámetro de Control | Inclusión del Límite Superior | Uso Principal |
| :--- | :--- | :--- | :--- |
| **`np.arange(inicio, fin, paso)`** | Controlas el **tamaño del salto** (`paso`). | **Excluye** `fin` (intervalo semiabierto `[inicio, fin)`). | Índices enteros, secuencias discretas con incremento fijo. |
| **`np.linspace(inicio, fin, num)`** | Controlas la **cantidad total de puntos** (`num`). | **Incluye** `fin` por defecto (`endpoint=True`). | Funciones continuas, gráficos científicos, divisiones exactas. |

### Sintaxis Mínima

````python
# 1. np.arange: Generar enteros pares del 10 al 38 (el 40 queda excluido)
vector_pares = np.arange(10, 40, 2)
print("Vector con arange (15 elementos):", vector_pares)
print("Shape:", vector_pares.shape)

# 2. np.linspace: Generar exactamente 11 muestras equidistantes entre 0 y 1 (incluyendo 0 y 1)
muestras_tiempo = np.linspace(0, 1, 11)
print("\nVector con linspace (11 muestras):", muestras_tiempo)

# 3. Combinando arange con reshape para construir matrices de prueba rápidamente
matriz_secuencial = np.arange(1, 17).reshape(4, 4)
print("\nMatriz 4x4 generada con arange + reshape:\n", matriz_secuencial)
````

:::{warning} 
**Regla:** Para secuencias con decimales, utiliza siempre `np.linspace()`. Reserva `np.arange()` exclusivamente para pasos enteros.
:::
