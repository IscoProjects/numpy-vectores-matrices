## 3.1 Cambio de Forma (`reshape`, `ravel`, `flatten`) y Transposición (`.T`, `np.transpose`)

Manipular la forma consiste en cambiar cómo NumPy interpreta los índices de un bloque de memoria sin alterar el orden secuencial de los datos almacenados.

### Funciones Principales
* **`arr.reshape(filas, columnas)`:** Cambia las dimensiones del arreglo manteniendo constante el número total de elementos (`size`). Puedes usar `-1` en una de las dimensiones para que NumPy calcule su tamaño automáticamente (ej. `arr.reshape(3, -1)`).
* **`arr.ravel()` vs. `arr.flatten()`:** Ambos aplana una matriz multidimensional a un vector 1D. La diferencia radica en la memoria: `ravel()` devuelve una **vista (*View*)** cuando es posible (más rápido), mientras que `flatten()` devuelve siempre una **copia (*Copy*)** independiente.
* **Transposición (`arr.T` o `np.transpose(arr)`):** Invierte los ejes de una matriz: una matriz de forma $(M, N)$ pasa a tener forma $(N, M)$, donde la fila $i$ se convierte en la columna $i$. Devuelve una **vista** en memoria.
* **`np.newaxis` (o `np.expand_dims`):** Inserta una dimensión de longitud `1` para convertir un vector 1D `(N,)` en un vector fila `(1, N)` o un vector columna `(N, 1)`.

### Sintaxis Mínima

````python
# 1. Vector 1D con 12 mediciones horarias
mediciones = np.arange(10, 130, 10)
print("Vector original (shape", mediciones.shape, "):", mediciones)

# 2. Reshape a matriz de 3 días (filas) x 4 turnos (columnas)
matriz_turnos = mediciones.reshape(3, 4)
print("\nMatriz 3x4 con reshape:\n", matriz_turnos)

# Uso de -1 para inferencia automática de columnas: 4 filas x N columnas
matriz_4_filas = mediciones.reshape(4, -1)
print("\nMatriz 4x3 (inferida con -1):\n", matriz_4_filas)

# 3. Transposición (.T y np.transpose son equivalentes en 2D)
transpuesta = matriz_turnos.T
print("\nMatriz transpuesta (shape", transpuesta.shape, "):\n", transpuesta)

# 4. Aplanar de 2D a 1D
vector_plano = matriz_turnos.flatten()
print("\nVector aplanado con flatten():", vector_plano)

# 5. Convertir un vector 1D (4,) en vector columna 2D (4, 1) para poder transponerlo
vector_1d = np.array([100, 200, 300, 400])
vector_columna = vector_1d[:, np.newaxis]
print("\nVector columna (shape", vector_columna.shape, "):\n", vector_columna)
print("Vector fila transpuesto (shape", vector_columna.T.shape, "):", vector_columna.T)
````

:::{note} Regla de Conservación del Tamaño (`size`)
Para que `arr.reshape(nueva_forma)` funcione, el producto de las nuevas dimensiones **debe ser exactamente igual** al `arr.size` original. Intentar redimensionar 12 elementos en una matriz de $3 \times 5$ (15 celdas) lanzará un `ValueError`.
:::

:::{warning} Trampa Común: Transponer un vector 1D con `.T`
Si tienes un vector 1D `v = np.array([1, 2, 3, 4])` con `shape = (4,)` y ejecutas `v.T`, **no ocurre absolutamente nada**: sigue teniendo `shape = (4,)`. Un arreglo de una sola dimensión no tiene segundo eje con el cual intercambiar posiciones. Para transponer un vector, primero debes promoverlo a 2D con `v[:, np.newaxis]` o `v.reshape(-1, 1)`.
:::

## 3.2 Concatenación de Arreglos (`np.concatenate`, `vstack`, `hstack`)

En proyectos reales, los datos suelen llegar fragmentados (por ejemplo, registros de distintos meses o nuevas variables medidas para los mismos sujetos). NumPy permite ensamblar arreglos a lo largo de un eje existente.

### Funciones Principales
* **`np.concatenate((arr1, arr2), axis=0)`:** Une los arreglos **verticalmente** (apila filas una debajo de otra). Requiere que ambos arreglos tengan el **mismo número de columnas**.
* **`np.concatenate((arr1, arr2), axis=1)`:** Une los arreglos **horizontalmente** (añade columnas una al lado de otra). Requiere que ambos arreglos tengan el **mismo número de filas**.
* **`np.vstack((arr1, arr2))` y `np.hstack((arr1, arr2))`:** Atajos semánticos (*Vertical Stack* y *Horizontal Stack*) que evitan recordar el número del eje y manejan automáticamente arreglos 1D.

### Sintaxis Mínima

````python
# Matriz A (3x4): Ventas de 3 tiendas en los primeros 4 meses (Ene-Abr)
ventas_t1 = np.array([
    [120, 135, 150, 140],
    [200, 210, 195, 220],
    [90,  105, 110, 115]
])
print("Matriz A (4x3)\n", ventas_t1)

# Matriz B (2x4): Ventas de 2 tiendas nuevas en los mismos 4 meses (Ene-Abr)
ventas_nuevas_tiendas = np.array([
    [160, 175, 180, 190],
    [110, 120, 125, 130]
])
print("\nMatriz B (2x4)\n", ventas_nuevas_tiendas)

# 1. Concatenación por FILAS (axis=0): Aumentan las filas (3 + 2 = 5 tiendas), columnas fijas (4)
todas_tiendas = np.concatenate((ventas_t1, ventas_nuevas_tiendas), axis=0)
print("\nConcatenación vertical axis=0 (shape", todas_tiendas.shape, "):\n", todas_tiendas)

# Matriz C (5x2): Ventas de las 5 tiendas en Mayo y Junio
ventas_may_jun = np.array([
    [155, 160],
    [230, 240],
    [120, 125],
    [195, 205],
    [135, 140]
])

# 2. Concatenación por COLUMNAS (axis=1): Filas fijas (5 tiendas), aumentan columnas (4 + 2 = 6 meses)
semestre_completo = np.concatenate((todas_tiendas, ventas_may_jun), axis=1)
print("\nConcatenación horizontal axis=1 (shape", semestre_completo.shape, "):\n", semestre_completo)

# 3. Equivalente usando hstack
semestre_hstack = np.hstack((todas_tiendas, ventas_may_jun))
print("\n¿np.concatenate(..., axis=1) es idéntico a np.hstack()?:", np.array_equal(semestre_completo, semestre_hstack))
````

:::{note} Regla de Alineación de Ejes en `np.concatenate`
Para concatenar a lo largo de un eje `k`, **todas las demás dimensiones deben coincidir exactamente**:
* Si `axis=0`: `(A, N)` + `(B, N)` $\rightarrow$ `(A + B, N)`.
* Si `axis=1`: `(M, A)` + `(M, B)` $\rightarrow$ `(M, A + B)`.
* Además, los arreglos a unir deben entregarse agrupados dentro de una **tupla o lista**: `np.concatenate((A, B), axis=0)`.
:::

:::{warning} Trampa Común: Intentar concatenar una matriz 2D con un vector 1D usando `np.concatenate`
Si tienes una matriz `M` de forma `(4, 3)` y un vector columna de datos `v = np.array([10, 20, 30, 40])` de forma `(4,)`, ejecutar `np.concatenate((M, v), axis=1)` fallará con `ValueError: all the input arrays must have same number of dimensions`.
**Solución:** Convierte primero `v` a 2D mediante `v[:, np.newaxis]` (forma `(4, 1)`) o usa `np.column_stack((M, v))`.
:::

## 3.3 Separación de Arreglos (`np.split`, `vsplit`, `hsplit`)

Es la operación inversa a la concatenación: permite dividir una matriz grande en una **lista de sub-arreglos** (por ejemplo, para separar variables predictoras `X` de la variable objetivo `y`, o dividir un lote de datos en bloques de entrenamiento y prueba).

### Dos Modos de División
1. **División en $N$ partes iguales:** Si pasas un número entero `N`, NumPy divide el eje en `N` submatrices del mismo tamaño.
2. **División por puntos de corte (índices):** Si pasas una lista de índices `[i, j]`, NumPy corta antes de cada índice indicado (generando los tramos `[:i]`, `[i:j]`, `[j:]`).

### Funciones Principales
* **`np.split(arr, secciones, axis=0)`:** Divide horizontalmente las filas (corta de arriba hacia abajo en bloques de filas). Equivalente a `np.vsplit(arr, secciones)`.
* **`np.split(arr, secciones, axis=1)`:** Divide verticalmente las columnas (separa grupos de columnas). Equivalente a `np.hsplit(arr, secciones)`.

### Sintaxis Mínima

````python
# Matriz 4x6: Dataset con 4 registros (filas) y 6 variables (columnas)
dataset = np.arange(1, 25).reshape(4, 6)
print("Dataset original (4x6):\n", dataset)

# 1. División por FILAS en 2 bloques iguales (axis=0: dos matrices de 2x6)
mitad_superior, mitad_inferior = np.split(dataset, 2, axis=0)
print("\nMitad superior (2x6):\n", mitad_superior)
print("\nMitad inferior (2x6):\n", mitad_inferior)

# 2. División por COLUMNAS en 3 bloques iguales (axis=1: tres matrices de 4x2)
bloque_1, bloque_2, bloque_3 = np.split(dataset, 3, axis=1)
print("\nPrimer bloque de columnas (4x2):\n", bloque_1)

# 3. División por ÍNDICES DE CORTE (ej. separar primeras 4 columnas como X, columna 4 como Y, columna 5 como Z)
X, Y, Z = np.split(dataset, [4, 5], axis=1)
print("\nX (shape", X.shape, "), Y (shape", Y.shape, "), Z (shape", Z.shape, ")")
print("Contenido de X (primeras 4 columnas):\n", X)
print("Contenido de Y (columna 4 como eje Y):\n", Y)
print("Contenido de Z (columna 5 como eje Z):\n", Z)
````

:::{warning} Trampa Común: División entera no exacta en `np.split`
Si pasas un número entero `N` como segundo argumento (`np.split(dataset, 3, axis=0)`), la longitud de ese eje **debe ser exactamente divisible por `N`**. Como `dataset` tiene 4 filas y 4 no es divisible entre 3, NumPy lanzará `ValueError: array split does not result in an equal division`.
*Si necesitas dividir un arreglo en bloques de tamaño desigual sin error, utiliza **`np.array_split()`**.*
:::

:::{tip} Nota sobre la Memoria en `split`
Todas las submatrices devueltas por `np.split()`, `vsplit()` y `hsplit()` son **vistas (*Views*)** del arreglo original. Si vas a modificar los fragmentos por separado, recuerda aplicar `.copy()`.
:::

## ✅ Resumen del Módulo 3
* Usa **`reshape()`** para cambiar dimensiones preservando el `size` total, y apóyate en `-1` para inferir una dimensión automáticamente.
* Para transponer un vector 1D, promuévelo primero a 2D mediante `vec[:, np.newaxis]`; en matrices 2D usa `.T` o `np.transpose()`.
